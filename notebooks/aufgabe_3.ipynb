```python
import pandas as pd

from pathlib import Path

root = Path.cwd() if (Path.cwd() / "data" / "train.csv").exists() else Path.cwd().parent
df = pd.read_csv(
    root / "data" / "train.csv",
    sep=";",
    quotechar="'",
    header=None,
    names=["label", "text"],
)

counts = df["label"].value_counts()
percentages = df["label"].value_counts(normalize=True) * 100

distribution = pd.DataFrame({
    "Anzahl": counts,
    "Anteil (%)": percentages.round(2),
})

print(f"Gesamtanzahl Artikel: {len(df)}")
print()
print("Klassenverteilung:")
print(distribution)

most = counts.idxmax()
least = counts.idxmin()

print()
print(
    f"Häufigste Kategorie: {most} — {counts[most]} Artikel "
    f"({percentages[most]:.2f} %)"
)
print(
    f"Seltenste Kategorie: {least} — {counts[least]} Artikel "
    f"({percentages[least]:.2f} %)"
)
```
